```python
import pandas as pd
```

```python
#1. Load and Read csv file
df=pd.read_csv("house_prices_practice.csv")
print(df.columns)
print(df)
```

```python
#2. slicing and indexing
print(df.loc[0:5,"Id":"YearBuilt"])
print(df.iloc[0:6,0:5])
```

```python
#3. Explore the data using .info() and .describe()
print("Information about dataset: ")
print(df.info())
print("Description about dataset: ")
print(df.describe())
```

```python
#4. Boolean indexing
print(df[df["YearBuilt"]>2000].loc[0:30,"Id":"YearBuilt"])
print(df[df["GarageCars"]==0][["Id","GarageCars","GrLivArea"]])
```

```python
#5. Handle missing values
print(df.isnull().sum())
```

```python
#6. use groupby on dataset columns
df.groupby("YearBuilt")["SalePrice"].mean()
```
