```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_absolute_error,r2_score

#Load the clean dataset
df=pd.read_csv('kc_house_preprocessed.csv')
df
```

```python
#To make the graph super clean select only 80 rows
df_sample=df.sample(n=80,random_state=42)

#Choose one simple feature: House size (in square feet)
x=df_sample.drop(columns='price')
y=df_sample['price']

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
print(f"Training set: {x_train.shape[0]} houses")
print(f"Test set: {x_test.shape[0]} houses")
```

```python
#Create our "Bendy line" pipeline (Degree 2)
poly_model=make_pipeline(PolynomialFeatures(degree=2),LinearRegression())

#Teach the robot by fitting the model on training data
poly_model.fit(x_train,y_train)

print("Bendy Line Robot (Polynomial Regression) trained successfully!")
```

```python
#PRedict house price and find accuracy
y_pred=poly_model.predict(x_test)
accuracy=r2_score(y_test,y_pred)
#Accuracy on trained data
y_pre=poly_model.predict(x_train)
accuracy1=r2_score(y_train,y_pre)
print(f"Accuracy Score for test data: {accuracy:,.2%}") 
print(f"Accuracy Score for train data: {accuracy1:,.2%}")
```
