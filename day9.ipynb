```python
import pandas as pd
```

```python
df=pd.read_csv("train.csv")
#First two rows
print(df.head(2))
#Print shape
print(df.shape)
#Display summary
print(df.info())
```

```python
#Print type of fare column
print(type(df["Fare"]))
#Print first 5 rows and Name,Sex,Age columns
print(df.loc[0:4,["Name","Sex","Age"]])
#Select rows from index 10 upto 15(excluded) and first 5 columns
print(df.iloc[10:15,0:5])
```

```python
#create a subset containing passengers who survived(Survived==1)
survived_pass=df[df["Survived"]==1]
print(survived_pass)
print(df[df["Pclass"]==1])
print(df[(df["Sex"]=="female") & (df["Survived"]==1)])
print(df[df["Fare"]>100])
```

```python
# count missing values,find age median and fill age nan with median
print(df.isnull().sum())
med=df["Age"].median()
df["Age"]=df["Age"].fillna(med)
print(df.isnull().sum())
```

```python
#drop any rows where Embarked column is missing/null and print count of missing values per col
df=df.dropna(subset=["Embarked"])
print(df.isnull().sum())
```

```python
print(df.groupby("Sex")["Age"].mean())
print(df.groupby("Pclass")["Fare"].mean())
```
