```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
```

```python
#1. Line plot :- is used for trend analysis
days=[1,2,3,4,5] #x-axis
speed=[10,25,45,70,95]
plt.plot(days,speed) # gives line plot
plt.show()
```

```python
days=[1,2,3,4,5] #x-axis
speed=[10,25,45,70,95] #y-axis
plt.figure(figsize=(7,4)) # height and width of figure in inches
plt.plot(days,speed,color='red',marker='o',linestyle='--',linewidth=5,label="Speed")
plt.title("Speed Over Days",fontsize=14,color="darkred") # for title of graph
plt.xlabel("Days of Training",fontsize=12) # for x label
plt.ylabel("Flight Speed (mph)",fontsize=12) # for y label
#To add a grid
plt.grid(True)
plt.legend() # to show label name given in .plot()
plt.show()
```

```python
#2. Bar Plot:- used to compare categories
treats=["Chocolates","Gummy Bears","Cookies","Ice Cream"]
counts=[15,8,22,12]
plt.figure(figsize=(7,4))
#Draw the bars
plt.bar(treats,counts,color="skyblue",edgecolor="darkblue")
#Add labels and titles
plt.title("Treats Eaten by Friends This Week",fontsize=14,color='teal')
plt.xlabel("Types of Treats",fontsize=12)
plt.ylabel("Total Eaten Counts",fontsize=12)
# Display the plot
plt.show()
```

```python
#3. Scatter Plot :- used to analyse relationships, clusters or patterns
east_coords=[2,4,1,5,3,2.5,4.5,1.5]
north_coord=[3,1,4,2,5,3.5,1.5,4.2]
plt.figure(figsize=(7,4))
plt.scatter(east_coords,north_coord,color="gold",marker="^",s=100,edgecolor="gold")
plt.title("Treasure Chest Locations on the Map",fontsize=14,color="orange")
plt.xlabel("East-West Coodinate",fontsize=12)
plt.ylabel("Nort-South Coodinate",fontsize=12)
plt.grid(True)
plt.show()
```

```python
#4. Histogram :- used for data distribution analysis

np.random.seed(42) # value to be used for random value generation so we get fixed and same random no. every time
candies=np.random.normal(loc=20,scale=4,size=30).astype(int)  # used for generation of values loc(mean),scale(std. deviation),size(no. of values)
df_candy=pd.DataFrame({"Candies":candies})
print(df_candy)

#Set Seaborn style theme
sns.set_theme(style="whitegrid")
plt.figure(figsize=(7,4))

#Draw the histogram with KDE (Kernel Density Estimation) line
sns.histplot(data=df_candy, x="Candies", kde=True, color="purple", bins=8) # data(for dataframe),x(for column of datafram),kde(for curve over bars of hist), bins(for no. of bars)

plt.title("Candy Distribution Among kids",fontsize=14)
plt.xlabel("Number of Candies",fontsize=12)
plt.ylabel("Number of Kids(Count)",fontsize=12)
plt.show()
```

```python
#5. Box Plot:- used for detection of outliers
frog_jumps=[5,6,4,7,5,7,6,7,5,6,5,22,18,20]
df_frogs=pd.DataFrame({"JumpDistance":frog_jumps})
plt.figure(figsize=(7,4))

sns.set_theme(style="whitegrid")
sns.boxplot(data=df_frogs,x="JumpDistance",color="lightgreen")
plt.title("Frog Jump Distances & Outlier Detection",fontsize=14,color="darkgreen")
plt.xlabel("Jump Distance (Feet)",fontsize=12)
plt.show()
```

```python
#6. Correlation Heatmap:- shows how strongly different numeric columns are related to each other(+ve(both directly related),zero(no relation),-ve(inverse relation))
df=pd.DataFrame({
    "PlayTime_Hours":[10,50,80,20,95],
    "Popularity_Score":[40,85,90,30,98],
    "Game_Difficulty":[8,5,3,9,2]
})
#Calculate correlation matrix
corr_matrix=df.corr()
print("Correlation Numbers:\n" , corr_matrix)
#Plot Heatmap
plt.figure(figsize=(7,4))
sns.heatmap(corr_matrix,annot=True,cmap="coolwarm",vmin=-1,vmax=1,linewidth=1) # annot(for no.s in boxes like .93,1 etc),cmap(for colors to be used coolwarm means red for +ve and blue for -ve),vmin(min. correlation vlaue),vmax(max value),linewidth(width between boxes)
plt.title("Video GAme Features Correlation Grid",fontsize=14)
plt.show()
```
