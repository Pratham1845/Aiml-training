```python
#Ridge lasso regularization
#We have 2 popular leashes to control our robot:- 1. Ridge Regression(it shrinks all weights of curve closer to zero but not make exactly zero) and 2. Lasso Regression(it shrinks some weights to exactly zero and hence can be used for feature selection). Both of these methods are used to prevent overfitting in regression models by adding a penalty term to the loss function.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_absolute_error,r2_score

# Load the dataset
df=pd.read_csv('kc_house_preprocessed.csv')
df.head()
```

```python
df_sample=df.sample(n=80,random_state=42)
#choose one scled feature:House Size(Scaled between 0 and 1)
x=df_sample[['Scaled_size']]
y=df_sample['price']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
print(f"Training set : {x_train.shape[0]} houses")
print(f"Test set : {x_test.shape[0]} houses")
```

```python
#Train Model of degree 6
wild_model=make_pipeline(PolynomialFeatures(degree=6),LinearRegression())
wild_model.fit(x_train,y_train)
#Prediction
y_pred_wild=wild_model.predict(x_test)
mae_wild=mean_absolute_error(y_test,y_pred_wild)
accuracy_wild=r2_score(y_test,y_pred_wild)
print(f"MAE Score: {mae_wild:,.2f} , Accuracy Score: {accuracy_wild:.2%}")
```

```python
#Train degree 6 model with ridge 
ridge_model=make_pipeline(PolynomialFeatures(degree=6),Ridge(alpha=0.1))
ridge_model.fit(x_train,y_train)

#Predict and calculate metrics
y_pred_ridge=ridge_model.predict(x_test)
mae_ridge=mean_absolute_error(y_test,y_pred_ridge)
accuracy_ridge=r2_score(y_test,y_pred_ridge)
print(mae_ridge,accuracy_ridge)
```

```python
#Strict Leash (Lasso)
lasso_model=make_pipeline(PolynomialFeatures(degree=6),Lasso(alpha=100.0,max_iter=10000))
lasso_model.fit(x_train,y_train)
y_pred_lasso=lasso_model.predict(x_test)
mae_lasso=mean_absolute_error(y_test,y_pred_lasso)
accuracy_lasso=r2_score(y_test,y_pred_lasso)
print(mae_lasso,accuracy_lasso)
```
