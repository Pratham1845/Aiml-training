```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("train.csv")
print(df.head())
```

```python
# Age Distribution((Histogram)
plt.figure(figsize=(7,4))
sns.histplot(data=df,x="Age",color="red",kde=True)
plt.title("Titanic Passengers Age Distribution",fontsize=14)
plt.xlabel("Age of Passenges",fontsize=12)
plt.ylabel("Number of Passenges",fontsize=12)
plt.show()
```

```python
#Count plot for Survived Column
plt.figure(figsize=(7,4))
sns.countplot(data=df,x="Survived",palette="Set2")# pallette for color and can use hue for subgroups
plt.title("Titanic Passengers Survival Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="Fare",color="lightblue")
plt.title("Ticket Fare Distribution & Outliers")
plt.show()
```

```python
#scatter plot age vs fare groupby survived
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x="Age",y="Fare",hue="Survived",palette="Set1")
plt.grid(True)
plt.show()
```

```python
#Correlation of Survived,Pclass,Age,Fare
cols=["Survived","Pclass","Age","Fare"]
corr=df[cols].corr()
sns.heatmap(data=corr,annot=True,vmax=1,vmin=-1,cmap="coolwarm")
```
