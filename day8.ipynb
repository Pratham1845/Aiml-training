```python
data=pd.read_csv('data.csv')
print(data)
print(data.shape)
print(data.dtypes)
print(data.info())
print(data.head())#by default first 5
print(data.tail())#by default last 5
```

```python
print(data.columns)
print(data.columns.to_list())#can also make numpy,series etc.. in same way
print(data.describe())
print(data['CustomerID'])
cols=['CustomerID','Age','Purchased']
print(data[cols])
```

```python
print(data.loc[0:3,'CustomerID':'Income'])#df.loc[rows range,col name range] include stop value also in range
cols=["Gender","Purchased"]
print(data.loc[3:4,cols])
print(df.iloc[0:3,0:3])#give index only of column also not column name and works upto stop-1
```

```python
#Boolean Indexing
print(data["Income"]>50000)
print(data[data["Income"]>50000])
print(data[data["Age"]>30])# to change dtype of column use data["Age"]=data["Age"].astype(int)
```

```python
print(data[(data["Income"]>=45000) & (data["Gender"]=="F")])
print(data.isnull())
print(data.isnull().sum())
mean=data["Age"].mean()
print(mean)
median=data["Age"].median()
print(median)
```

```python
data["Age"]=data["Age"].fillna(mean)
print(data.isnull().sum())
data=data.dropna(subset=["Gender"])#if drop all columns NAN row then .dropna()
print(data)
```

```python
print(data.groupby("Gender")["Age"].mean())
print(data.groupby("Gender")["Income"].mean())
print(data.groupby("Gender")[["Age","Income"]].mean())
print(data.groupby("Purchased")[["Age","Income"]].mean())
```
