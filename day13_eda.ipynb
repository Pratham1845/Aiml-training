```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("kc_house_data.csv")
print(df.columns.to_list())
df.head()
```

```python
cols=["price","bedrooms","bathrooms","sqft_living","floors","waterfront","yr_built","condition"]
df=df[cols]
df.head()
```

```python
#Data distribution for price
plt.figure(figsize=(10,5))
sns.histplot(data=df,x="price",color="red",bins=50,kde=True)
plt.title("The Price Mountain: How House Prices are Distributed",fontsize=14)
plt.xlabel("Price ($)",fontsize=12)
plt.ylabel("Number of Houses",fontsize=12)
plt.show()
```

```python
# Count plot for bedrooms
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="bedrooms",hue="bedrooms",palette="Set2",legend=False)# to show differnt colors for each bar or bedrooms we use hue and palette
plt.show()
```

```python
#Boxplot for bedrooms for outliers
plt.figure(figsize=(8,5))
sns.boxplot(data=df,x="bedrooms",color='lightblue')
plt.show()
```

```python
#Relation between Size and price and groupby waterfront
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x="sqft_living",y="price",hue="waterfront",color="coral")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df,x="waterfront",y="price",hue="waterfront",palette="pastel")
plt.title("Waterfront View vs Price: Is lake-side living expensive?",fontsize=14)
plt.xticks([0,1],['No Waterfront','Waterfront'])
plt.xlabel("House Location",fontsize=12)
plt.ylabel("Price ($)",fontsize=12)
plt.ticklabel_format(style="plain",axis="y")# to show price values as it is
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.barplot(data=df,x="condition",y="price",hue="condition",palette="coolwarm")
plt.title("House Condition vs Price: Do cleaner houses sell for more?",fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(10,8))
correlation_map= df.corr()
sns.heatmap(correlation_map,annot=True,cmap="coolwarm",linewidths=0.5,vmin=-1,vmax=1)
plt.title("The Magic Connection Map (Correlation Heatmap)",fontsize=16)
plt.show()
```
